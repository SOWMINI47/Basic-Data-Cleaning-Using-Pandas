```python
import pandas as pd

# 1. Create Dataset
data = {
    "Student_ID": [101, 102, 103, 104, 104, 105, 106],
    "Name": ["Anu", "Bala", "Charan", "Divya", "Divya", "Esha", None],
    "Age": [20, 21, None, 22, 22, 19, 20],
    "Marks": [85, 90, 78, None, None, 88, 92],
    "Attendance": ["95", "88", "90", "85", "85", "92", "91"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# 2. Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Check Duplicate Rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# 4. Remove Duplicate Rows
df_clean = df.drop_duplicates().copy()

# 5. Handle Missing Values
df_clean["Name"] = df_clean["Name"].fillna("Unknown")
df_clean["Age"] = df_clean["Age"].fillna(
    df_clean["Age"].median()
)
df_clean["Marks"] = df_clean["Marks"].fillna(
    df_clean["Marks"].mean()
)

# 6. Correct Data Types
df_clean["Age"] = df_clean["Age"].astype(int)
df_clean["Marks"] = df_clean["Marks"].round(2)
df_clean["Attendance"] = pd.to_numeric(
    df_clean["Attendance"], errors="coerce"
).astype("Int64")

# 7. Display Cleaned Dataset
print("\nCleaned Dataset:")
print(df_clean)

# 8. Verify Missing Values
print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())

# 9. Verify Duplicate Rows
print("\nDuplicate Rows After Cleaning:")
print(df_clean.duplicated().sum())

# 10. Save Cleaned Dataset
df_clean.to_csv("cleaned_student_dataset.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
